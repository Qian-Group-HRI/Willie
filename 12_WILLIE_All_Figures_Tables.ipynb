# WILLIE — Complete Paper Figures & Tables
Generates ALL data-driven figures and ALL tables as publication-ready PNGs.
Output: `paper_figs/` (figures) and `paper_tables/` (tables as images)

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import os, warnings
warnings.filterwarnings('ignore')

plt.rcParams.update({'font.family':'serif','font.serif':['Times New Roman','DejaVu Serif'],
    'font.size':11,'axes.titlesize':13,'axes.labelsize':11,'xtick.labelsize':10,
    'ytick.labelsize':10,'legend.fontsize':9,'figure.dpi':300,'savefig.dpi':300,
    'savefig.bbox':'tight','savefig.pad_inches':0.05,'axes.grid':False,
    'axes.spines.top':False,'axes.spines.right':False})

C_BLUE='#2563EB'; C_RED='#DC2626'; C_GREEN='#16A34A'; C_ORANGE='#EA580C'
C_PURPLE='#9333EA'; C_TEAL='#0D9488'; C_GRAY='#6B7280'; C_DARK='#1F2937'
WOUND_COLORS=['#E74C3C','#F39C12','#3498DB','#9B59B6','#2ECC71']

FIG_DIR='paper_figs'; TBL_DIR='paper_tables'
os.makedirs(FIG_DIR, exist_ok=True); os.makedirs(TBL_DIR, exist_ok=True)

def savefig(fig, name, folder=FIG_DIR):
    path=os.path.join(folder, f'{name}.png')
    fig.savefig(path, dpi=300, bbox_inches='tight', facecolor='white')
    plt.close(fig); print(f'  Saved: {path}')

def render_table(data, col_labels, row_labels=None, title='', name='', col_widths=None, highlight_row=None, highlight_col=None):
    """Render a table as a clean PNG image."""
    n_rows = len(data); n_cols = len(col_labels)
    fig_w = max(8, n_cols * 1.4); fig_h = max(2, (n_rows+1)*0.45 + 0.8)
    fig, ax = plt.subplots(figsize=(fig_w, fig_h))
    ax.axis('off')
    if title: ax.set_title(title, fontsize=12, fontweight='bold', pad=10)
    cell_text = [[str(c) for c in row] for row in data]
    if row_labels:
        cell_text = [[rl]+row for rl, row in zip(row_labels, cell_text)]
        col_labels = ['']+list(col_labels)
    tbl = ax.table(cellText=cell_text, colLabels=col_labels, loc='center', cellLoc='center')
    tbl.auto_set_font_size(False); tbl.set_fontsize(10); tbl.scale(1, 1.6)
    for (r,c), cell in tbl.get_celld().items():
        if r==0: cell.set_facecolor('#1565C0'); cell.set_text_props(color='white', fontweight='bold')
        elif r%2==0: cell.set_facecolor('#F5F5F5')
        else: cell.set_facecolor('white')
        cell.set_edgecolor('#DDDDDD')
        if highlight_row is not None and r==highlight_row+1:
            cell.set_facecolor('#E8F5E9'); cell.set_text_props(fontweight='bold')
    savefig(fig, name, TBL_DIR)

print('Setup complete.')
```
